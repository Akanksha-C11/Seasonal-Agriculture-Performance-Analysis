**7. Outlier Analysis**

```python
#select important numerical columns
outlier_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Revenue_INR',
    'Profit_INR',
    'Water_Used_m3'
]
```

```python
#boxplots for outlier visualization
plt.figure(figsize=(15, 10))
for i, column in enumerate(outlier_columns, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python
#identify outliers using IQR method
for column in outlier_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(column, ":", len(outliers), "outliers")
```

Outliers were investigated using boxplots and the IQR method for important
numerical variables such as rainfall, temperature, yield, production, revenue,
profit, and water usage.
Potential outliers were identified based on values falling below the lower limit
or above the upper limit calculated using the IQR method.
The identified outliers were retained in the dataset because extreme agricultural
or economic values may represent genuine real-world variations rather than data
errors. Therefore, they may provide meaningful insights for the analysis.

```python
#create profit status column
df['Profit_Status'] = df['Profit_INR'].apply(
    lambda x: 'Profit' if x >= 0 else 'Loss'
)
#count profit and loss records
df['Profit_Status'].value_counts()
```

```python
#loss percentage by season
loss_percentage = (
    df[df['Profit_Status'] == 'Loss']
    .groupby('Season')
    .size()
    / df.groupby('Season').size()
) * 100
loss_percentage
```

The analysis compared profit and loss records across different seasons and
calculated the percentage of loss-making farms in each season.
This helps identify seasonal differences in the economic performance of
agricultural activities.
